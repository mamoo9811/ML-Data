```python
import pandas as pd
fname = "/content/drive/MyDrive/AXSOSACADEMY/01-Fundamentals/Week02/Data/bake_sale_data.csv"
df = pd.read_csv(fname)
df.head()
```

```python
df.value_counts()
```

```python
df["Student"].value_counts()
```

```python
df["Student"].value_counts(normalize=True)
```

```python
columns = df.columns
filter_quantity_cols = columns[columns.str.contains('Quantity')]
filter_quantity_cols
```

```python
df[filter_quantity_cols]
```

```python
df[filter_quantity_cols].sum(axis=0)
```

```python
df[filter_quantity_cols].sum(axis=1)
```

```python
df['Comulative Quantity (Day 4)'] = df[filter_quantity_cols].sum(axis=1)
```

```python
df.head()
```

```python
df['Comulative Quantity (Day 4)'].sort_values(ascending=False)
```

```python
df['total_profits'] = (df['Comulative Quantity (Day 4)'] * df['Price'] ) - (df['Comulative Quantity (Day 4)'] * df['Expense'])
```

```python
df.head()
```

```python
df['total_profits'].sort_values(ascending=False)
```

```python
df['Student'].value_counts().max()
```

```python
student_sums = df.groupby("Student").sum()
student_sums
```

```python
student_sumss = df.groupby("Student")["Comulative Quantity (Day 4)"].sum()
student_sumss.sort_values(ascending=False)
```

```python
most_total_profit = df.groupby("Student")['total_profits'].sum()
most_total_profit.sort_values(ascending=False)
```
