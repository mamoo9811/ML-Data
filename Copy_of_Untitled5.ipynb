```python
nums1 = [1, 2, 3, 0, 0, 0]
m = 3
nums2 = [2, 5, 6]
n = 3
```

```python
import pandas as pd

def merge(nums1, m, nums2, n):
    # ناخد بس العناصر الحقيقية من كل مصفوفة
    real_nums1 = nums1[:m]
    real_nums2 = nums2[:n]

    # ندمجهم بـ Series واحدة باستخدام pd.concat
    combined = pd.concat([pd.Series(real_nums1), pd.Series(real_nums2)])

    # نرتبها تصاعديًا
    sorted_combined = combined.sort_values()

    # نحولها لقائمة ونحطها مكان nums1
    result = sorted_combined.tolist()

    for i in range(len(result)):
        nums1[i] = result[i]

    return nums1
```

```python
print(merge(nums1, m, nums2, n))
```
