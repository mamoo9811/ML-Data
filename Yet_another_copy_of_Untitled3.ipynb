```python
import pandas as pd
```

```python
df1 = pd.DataFrame(
    { 'A' : [1, 2, 3]},
index=['a','b','c']
)

df2 = pd.DataFrame(
    {'B' : [4, 5, 6]},
    index=['a', 'b', 'd']
)

result = df1.join(df2, how='outer')
print(result)
```

```python
emp_N = pd.DataFrame(
    {'name' : ['ahmad', 'sara' , 'omar']},
    index=['E1','E2','E3']
)

emp_S = pd.DataFrame(
    {'salary' : [3000, 3500 ]},
    index=['E1','E2']
)

result2 = emp_N.join(emp_S, how='outer')
```

```python
print(result2)
```

```python
cust_N = pd.DataFrame(
    {'customer_ID': [1, 2, 3],
    'Name': ['Nour', 'Ahmad', 'Lina']}
)
cust_O = pd.DataFrame(
    {'customer_ID': [1, 3],
    'Order': ['laptop', 'mouse']}
)
customers = pd.merge(cust_N, cust_O, on='customer_ID')
```

```python
print(customers)
```
