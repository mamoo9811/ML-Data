```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

filename = "/content/drive/MyDrive/AXSOSACADEMY/01-Fundamentals/Week04/insurance_mod.csv"
df = pd.read_csv(filename)
pd.set_option('display.max_columns',50)
df.head()
```

```python
df.info()
```

# **Explore correlations**

```python
fix ,ax =plt.subplots()
corr = df.corr(numeric_only=True)
sns.heatmap(corr, cmap = 'coolwarm', annot=True,ax=ax);
plt.title('Correlation Heatmap')
#Q1/1 -> yes there is a smoker vs charges
#Q1.2 -> its a positive
```

# **Explore smoker vs charges**


1:Barplot

```python
sns.barplot(data=df, x='smoker', y='charges')
plt.title('Average Charges for Smokers vs. Non-Smokers')
#Q2 -> its higher for smokers than for non-smokers
```

2:Boxplot

```python
sns.boxplot(data=df, x='smoker', y='charges')
plt.title('Charges for Smokers vs. Non-Smokers')
#Q3 -> the median, data distribution, variability and outliers
```

9:Regplot – Charges vs. BMI

```python
line_kws =dict(color='black')
scot_kws =dict(edgecolor='white')
ax = sns.regplot(data=df, x='bmi', y='charges', line_kws= line_kws, scatter_kws=scot_kws);
#Q9 -> There is a positive relationship
```

10:Scatterplot – BMI colored by Smoker

```python
sns.scatterplot(data=df, x='bmi', y='charges', hue='smoker')
plt.title('Charges vs BMI by Smoker')
#Q10 -> Smokers have higher charges than non-smokers
```

11:Linear Model Plot – BMI colored by Smoker

```python
g = sns.lmplot(data=df, x='bmi', y='charges', hue='smoker',scatter_kws=scot_kws);
plt.title('Charges vs BMI by Smoker')
#Q11 ->   the BMI and charges have a positive relationship
```

12:Countplot for Regions

```python
sns.countplot(data=df, x='region')
plt.title('Number of Customers by Region')
#Q12 -> the Southeast has the highest number of customers
```

13:Barplot of Charges by Region

```python
sns.barplot(data=df, x='region', y='charges')
plt.title('Average Charges by Region')
#Q13 -> the Southeast has the highes charges
```

14:Bar Chart of the Ratios of Smokers in Each Region

```python
percent_smo = df.groupby('region')['smoker'].mean()
ind = (percent_smo.index)
val = (percent_smo.values)
sns.barplot(x=ind, y=val)
plt.title('Percentage of Smokers by Region')
#Q14 -> the Southeast has the highest ratio of smokers
```
